### P1 — Regression

```python
from mibici.modeling.p1 import load_p1_data
from mibici.features import build_p1_features

feature_data = build_p1_features(load_p1_data())

train_data = feature_data.loc[
    feature_data["date"] < "2026-09-01"
].copy()
```

```python
import matplotlib.pyplot as plt

train_data["departures"].hist(bins=50, figsize=(8, 4))

plt.title("Salidas diarias por estación — entrenamiento")
plt.xlabel("Número de salidas")
plt.ylabel("Cantidad de filas estación–día")
plt.tight_layout()
plt.show()
```

El objetivo presenta asimetría positiva y una cola de conteos altos.
Se conserva su escala original para la primera regresión lineal.
No se recortan valores únicamente por la forma de la distribución.
Una transformación log1p podrá evaluarse como experimento posterior.
