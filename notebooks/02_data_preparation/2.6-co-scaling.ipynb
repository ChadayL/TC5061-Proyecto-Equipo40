### P1 — Regression

```python
from mibici.modeling.p1 import load_p1_data
from mibici.features import build_p1_features
import pandas as pd

feature_data = build_p1_features(load_p1_data())

train_data = feature_data.loc[
    feature_data["date"] < "2026-09-01"
].copy()

validation_data = feature_data.loc[
    feature_data["date"] >= "2026-09-01"
].copy()
```

```python
numeric_features = [
    "departures_lag_1",
    "departures_lag_7",
    "departures_mean_7",
]
```

```python
from mibici.data_preparation.scaling import build_scaler

scaler = build_scaler()

scaled_train = scaler.fit_transform(
    train_data[numeric_features]
)

scaled_validation = scaler.transform(
    validation_data[numeric_features]
)
```

```python
import pandas as pd

scaled_train_df = pd.DataFrame(
    scaled_train,
    columns=numeric_features,
    index=train_data.index,
)

display(
    scaled_train_df.agg(["mean", "std"])
)
display(scaled_train_df.std(ddof=0))
```

StandardScaler se ajustó exclusivamente con entrenamiento.
Los predictores escalados presentan media cercana a cero y
desviación estándar unitaria. Validación se transformó sin reajuste.
Los faltantes se conservan.
