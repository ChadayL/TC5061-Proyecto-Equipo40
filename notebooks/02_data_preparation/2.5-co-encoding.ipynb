### P1 — Regression

```python
from mibici.modeling.p1 import load_p1_data
from mibici.features import build_p1_features

feature_data = build_p1_features(load_p1_data())

train_data = feature_data.loc[
    feature_data["date"] < "2026-09-01"
].copy()

validation_data = feature_data.loc[
    feature_data["date"] >= "2026-09-01"
].copy()
```

```python
categorical_features = ["station_id", "day_of_week"]
```

```python
from mibici.data_preparation.encoding import build_encoder

categorical_features = ["station_id", "day_of_week"]

encoder = build_encoder()

encoded_train = encoder.fit_transform(
    train_data[categorical_features]
)

encoded_validation = encoder.transform(
    validation_data[categorical_features]
)
```

```python
assert encoded_train.shape[1] == encoded_validation.shape[1]

print(f"Entrenamiento: {encoded_train.shape}")
print(f"Validación: {encoded_validation.shape}")

display(encoder.get_feature_names_out(categorical_features))
```

```python
for column, known_categories in zip(
    categorical_features,
    encoder.categories_,
):
    unknown_values = validation_data.loc[
        ~validation_data[column].isin(known_categories),
        column,
    ]

    print(f"{column}:")
    print(f"  Categorías nuevas: {unknown_values.unique().tolist()}")
    print(f"  Filas afectadas: {len(unknown_values):,}")
```

El encoder se ajustó únicamente con entrenamiento y produjo
477 columnas compatibles en ambos conjuntos.

Se detectó una estación desconocida en validación: ID 180,
con 20 filas. Se procesa mediante handle_unknown="ignore".
