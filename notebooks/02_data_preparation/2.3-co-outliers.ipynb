```python
from mibici.modeling.p1 import load_p1_data
from mibici.features import build_p1_features
import pandas as pd

daily_demand = load_p1_data()
feature_data = build_p1_features(daily_demand)

train_data = feature_data.loc[
    feature_data["date"] < "2026-09-01"
].copy()

validation_data = feature_data.loc[
    feature_data["date"] >= "2026-09-01"
].copy()
```

```python
station_train_summary = (
    train_data.groupby("station_id")["departures"]
    .agg(["count", "median", "mean", "max"])
    .sort_values("max", ascending=False)
)

display(station_train_summary.head(15))
```

```python
import pandas as pd

for station_id in [51, 396, 196]:
    station_data = (
        train_data.loc[train_data["station_id"].eq(station_id)]
        .sort_values("date")
    )

    peak_date = station_data.loc[
        station_data["departures"].idxmax(), "date"
    ]

    print(f"\nEstación {station_id}")

    display(
        station_data.loc[
            station_data["date"].between(
                peak_date - pd.Timedelta(days=3),
                peak_date + pd.Timedelta(days=3),
            ),
            ["station_id", "date", "departures", "day_of_week"],
        ]
    )
```
