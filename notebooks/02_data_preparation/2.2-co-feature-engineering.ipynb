```python
from mibici.modeling.p1 import load_p1_data
from mibici.features import (
    add_calendar_features,
    add_lag_features,
    add_rolling_features,
)

daily_demand = load_p1_data()

feature_data = add_calendar_features(daily_demand)
feature_data = add_lag_features(feature_data)
feature_data = add_rolling_features(feature_data)

display(feature_data.head(10))
```

```python
display(
    feature_data[
        ["departures_lag_1", "departures_lag_7", "departures_mean_7"]
    ].isna().sum()
)
```

```python
required_features = [
    "departures_lag_1",
    "departures_lag_7",
    "departures_mean_7",
]

model_data = feature_data.dropna(
    subset=required_features
).copy()

print(f"Filas disponibles para modelado: {len(model_data):,}")
```

Se implementaron día de la semana, rezagos de 1 y 7 días y
promedio de los 7 días anteriores.

Se verificó la conservación de filas y del objetivo.
Los faltantes iniciales corresponden a historia no disponible
y se conservan. Las variables temporales excluyen el día predicho.
